```python
import os
import pickle

# load chunked dataset
chunks = []
PREPROCESSED_DATASET_PATH = os.path.join("../artifacts" , "chunked_dataset.pkl")
with open(PREPROCESSED_DATASET_PATH, 'rb') as file:
    chunks = pickle.load(file, encoding='utf-8')

if not chunks:
    print("chunks not loaded.")

print(chunks[0].page_content)
```

```python
# create vector embeddings 
from langchain_ollama.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

embeddings = OllamaEmbeddings(model="nomic-embed-text")

vector_store = FAISS.from_documents(chunks, embeddings)

# VECTOR_EMBEDDINGS_PATH = os.path.join(os.getcwd(), "artifacts", "vectore_store.pkl")
VECTOR_EMBEDDINGS_PATH = os.path.join("../artifacts", "vector_store")
vector_store.save_local(VECTOR_EMBEDDINGS_PATH)
```

```python
# final model
from langchain_ollama import ChatOllama
llm = ChatOllama(model="mistral:latest")

query = "Tell me about IPC section 375?"
docs = vector_store.similarity_search_with_score(query)

context = "\n\n".join([d.page_content for d, s in docs])

response = llm.invoke(f"Answer using context:\n\n{context}\n\nQuestion: {query}")

print(response.content)
```
